```python
for frame_name, df in dataframes.items():
    for col in to_date_columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')
```

## 1.1 MoM/YoY Growth

```python
MoM_growth = sales_transactions.pivot_table(
    index='Month_Key', values=['Gross_Sales_EGP'],
    aggfunc='sum').sort_index()

MoM_growth['Gross_Sales_EGP_Growth_Pct'] = round(MoM_growth['Gross_Sales_EGP'].pct_change() * 100, 2) 

agg_cols = ['Net_Sales_EGP', 'Estimated_COGS_EGP', 'Gross_Profit_EGP', 'Refund_Value_EGP', 'Quantity']
for col in agg_cols:
    agg_index='Month_Key'
    MoM_growth[col] = sales_transactions.pivot_table(
                                    index=agg_index, values=[col],
                                    aggfunc='sum').sort_index()[col]
    MoM_growth[col] = round(MoM_growth[col], 2)
    if col == 'Net_Sales_EGP':
        MoM_growth[f'{col}_Growth_Pct'] = round(MoM_growth[col].pct_change() * 100, 2)
    if col == 'Gross_Profit_EGP':
        MoM_growth['Gross_Margin_Pct'] = round(MoM_growth['Gross_Profit_EGP'] / MoM_growth['Net_Sales_EGP'] * 100, 2)

MoM_growth.reset_index(inplace=True)
MoM_growth.head()
```

```python
# we have 36 months we need the YoY To split them into 3 years 12 months each 
First_year = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2023-10-1') &
    (sales_transactions['Order_Date'] <= '2024-09-30')]
First_year['12_month_period'] = 'period 1'

Second_year = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2024-10-1') &
    (sales_transactions['Order_Date'] <= '2025-09-30')]
Second_year['12_month_period'] = 'period 2'

Third_year = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2025-10-1') &
    (sales_transactions['Order_Date'] <= '2026-09-30')]

Third_year['12_month_period'] = 'period 3'

All_Three_Years = pd.concat([First_year, Second_year, Third_year], ignore_index=True)
All_Three_Years.head()
```

```python
plot_sales = sales_transactions.pivot_table(
    index='Month_Key', values=['Gross_Sales_EGP'],
    aggfunc='sum').sort_index()

plot_sales_year1 = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2023-10-1') &
    (sales_transactions['Order_Date'] <= '2024-09-30')].pivot_table(
        index='Month_Key', values=['Gross_Sales_EGP'],
        aggfunc='sum').sort_index()

plot_sales_year2 = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2024-10-1') &
    (sales_transactions['Order_Date'] <= '2025-09-30')].pivot_table(
        index='Month_Key', values=['Gross_Sales_EGP'],
        aggfunc='sum').sort_index()

plot_sales_year3 = sales_transactions.loc[
    (sales_transactions['Order_Date'] >= '2025-10-1') &
    (sales_transactions['Order_Date'] <= '2026-09-30')].pivot_table(
        index='Month_Key', values=['Gross_Sales_EGP'],
        aggfunc='sum').sort_index()

plt.figure(figsize=(32, 8))
plt.plot(plot_sales.index, plot_sales['Gross_Sales_EGP'], marker='o', color='blue', label='Gross Sales EGP')
plt.plot(plot_sales_year1.index, plot_sales_year1['Gross_Sales_EGP'], marker='s', color='red', label='Gross Sales EGP (2023-2024)')
plt.plot(plot_sales_year2.index, plot_sales_year2['Gross_Sales_EGP'], marker='^', color='green', label='Gross Sales EGP (2024-2025)')
plt.plot(plot_sales_year3.index, plot_sales_year3['Gross_Sales_EGP'], marker='*', color='orange', label='Gross Sales EGP (2025-2026)')
plt.title('Gross Sales Over Time')
plt.xlabel('Month Key')
plt.ylabel('Gross Sales (EGP)')
plt.legend()
plt.show()
```
