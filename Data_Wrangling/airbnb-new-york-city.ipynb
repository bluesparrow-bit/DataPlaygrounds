# Audit missing data
Report how many missing values exist in each column, and what percentage of rows are affected overall. Columns that are 100% empty should be identified separately from partially-missing ones.

PS: `neighbourhoods.csv` had no missing Values at the time this analysis was made

```python
col_missing_count = listings.isna().sum()[listings.isna().sum() != 0]
col_missing_count.sort_values(ascending=False, inplace=True)
col_missing_count
```

```python
col_missing_pct = (listings.isna().mean()[listings.isna().mean() != 0]) * 100

col_missing_pct.sort_values(ascending=False, inplace=True)

col_missing_pct = round(col_missing_pct,2)
fully_empty_columns = col_missing_pct[col_missing_pct == 100]
partially_missing_columns = col_missing_pct[col_missing_pct < 100]
# Notice that we used the missing values percent per column because
# when the data increases in the futrue this won't get affected but if we
# used the total missing count (30259 now) this would change when data size increases

print("Fully empty columns:")
print(fully_empty_columns)

print("\nPartially missing columns:")
print(partially_missing_columns)
```

```python
rows_missing_count = listings.isna().any(axis=1).sum()

rows_missing_pct = listings.isna().any(axis=1).mean() * 100

print(f"Rows with at least one missing value: {rows_missing_count:,}")
print(f"Overall percentage of affected rows:  {rows_missing_pct:.2f}%")
```

# Clean the price column
`price` is stored as text with a leading "$" (e.g. "$113.97"). Strip the currency symbol and commas, then cast to `float` so it can be used numerically.

```python
listings['price'] = (
    listings['price']
    .str.replace('$', '')
    .str.replace(',', ''))
listings.price.head(3)
```

```python
listings.price = listings.price.astype('float64')
listings.price.dtype
```

# Fill numeric gaps	
Fill missing `bedrooms`, `bathrooms`, and `beds` using the `median` value for that listing's `room_type` and `accommodates` group (not the overall `median`).

```python
group_cols = ['room_type','accommodates']

numeric_cols = ['bedrooms','bathrooms','beds']

print(listings[numeric_cols].isna().sum())
print(t)
print(listings.dtypes[numeric_cols])
```

```python
for col in numeric_cols:
    group_median = listings.groupby(group_cols)[col].transform('median')
    listings.fillna({col:group_median}, inplace=True)
print(listings[numeric_cols].isna().sum())
```

```python
# First-pass logic is correct. but we still have missing values. 
# For these exceptional cases, use a documented fallback: the median for the same room_type alone.
for col in numeric_cols:
    group_median = listings.groupby('room_type')[col].transform('median')
    listings.fillna({col:group_median}, inplace=True)
print(listings[numeric_cols].isna().sum())
```

# Handle the license column
`license` is only ~82% populated. A missing license may be meaningful (unregistered listing) rather than a data error — add a boolean `has_license` column instead of dropping or imputing rows.
