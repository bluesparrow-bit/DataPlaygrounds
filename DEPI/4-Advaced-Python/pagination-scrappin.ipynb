```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC
from IPython.display import Image
```

```python
driver = webdriver.Firefox()

wait = WebDriverWait(driver, 10)

driver.set_window_size(1050, 1050)

driver.get("https://webscraper.io/test-sites/pagination")
```

```python

'''Fields on the listing pages
Title	
Tagline	
Year	
Country of origin	
Mileage	
Price	
Status	
Product URL	
Thumbnail URL
'''

RATING_MAP = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}

car = driver.find_element(By.CSS_SELECTOR, "div.card.sitemap-card.test-sites-card")
title = car.find_element(By.CLASS_NAME, "title")
status = car.find_element(By.CLASS_NAME, "badge").text.strip()
tagline = car.find_element(By.CLASS_NAME, "card-text").text.strip()
print("Status:", status)
print("Tagline:", tagline)
```
