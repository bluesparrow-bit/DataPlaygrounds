## Hadoop

### Run Hadoop on Docker

```python
docker exec -it master1 bash
```

### Copy File In Docker Container

```python
docker cp <Path>/FolderName <ContainerName>:<PATH>/<FileName>
```

### Check the instruction of a command

```python
hdfs dfs -help
```

### Display the directory information

```python
hdfs dfs -ls /
hdfs dfs -ls /<PATH>
```

### Hadoop HDFS Create Folder

```python
hdfs dfs -mkdir /<Path>/FolderName
```

### Hadoop HDFS Delete Folder or File

```python
** Delete Empty Folder or file **

hdfs dfs -rm /<PATH>/<FolderName or FileName>

** Delete Folder Contains Files **

hdfs dfs -rm -r /<PATH>/<FolderName>
```

### Upload a  local file to specified  directory in HDFS

```python
hdfs dfs -put /<SourcePATH>/<FileName> /<TargetPATH>
```

### Download  specified  file  from  HDFS to a  local host

```python
hdfs dfs -get /<SourcePATH>/<FileName> /<TargetPATH>
```

### Set Replica for file already on HDFS

```python
docker exec master1 hdfs dfs -setrep -w 3 /<SourcePATH>/<FileName>
```

### Set Replica for file while uploading

```python
hdfs dfs -D dfs.replication=3 -put -f /<SourcePATH>/<FileName> /<TargetPATH>
```

### Copy files from source to destination inside hdfs

```python
hdfs dfs -cp /<SourcePATH>/<FileName> /<TargetPATH>
```

### Move files

```python
hdfs dfs -mv /<SourcePATH>/<FileName> /<TargetPATH>/<FileName>
```

### Display the content of the  file

```python
hdfs dfs -cat /<PATH>/<FileName>
```

### Add data to the end of a file

```python
hdfs dfs -appendToFile /<SourcePATH>/<FileName> /<TargetPATH>/<TargetFileName>
```

### Check the available space of file system

```python
hdfs dfs -df
```

### Modify the permissions

```python
hdfs dfs -chmod 777 /<PATH>/<FolderName or FileName>
```

### See File Blocks

```python
hdfs fsck /<PATH>/<FileName> -files -blocks -locations
```
