```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/Titanic_420.csv')
print(df.info())
```

```python
print(df.isna().sum())
```

```python
print(df.isna().mean() * 100)
```

```python
sns.set_style('whitegrid')
fig, axes = plt.subplots(1, 3, figsize=(16,4)) 
sns.barplot(data=df,y='Sex',x='Survived',ax=axes[0],color='blue')
axes[0].set_title('Survival Rate by Sex')
sns.barplot(data=df,x='Pclass',y='Survived',ax=axes[1],color='green')
axes[1].set_title('Survival Rate by Passenger Class')
sns.barplot(data=df,x='Embarked',y='Survived',ax=axes[2],color='red')
axes[2].set_title('Survival Rate by Port of Embarkation')
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['Age'].dropna(), bins=20, kde=True, color='steelblue')
plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(visible=True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='Pclass', hue='Pclass',y='Fare', palette='Set3')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.grid(visible=True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', palette='coolwarm', alpha=0.7)
plt.title('Age vs Fare (colored by Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')
plt.grid(visible=True)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
