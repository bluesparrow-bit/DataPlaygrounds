### Equivalent Function Calls

In Python, equivalent function calls are different ways to call a function that produce the same result. This is possible because Python allows arguments to be passed in several ways:

- Positional arguments
- Keyword arguments
- Default arguments

```python
def describe_pet(pet_name, animal_type='dog'):
    print(f"I have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")
```

```python
# A dog named max.
describe_pet('max')
describe_pet(pet_name='max')
```

```python
# A cat named lilly.
describe_pet('lilly', 'cat')
describe_pet(pet_name='lilly', animal_type='cat')
describe_pet(animal_type='cat', pet_name='lilly')
describe_pet(pet_name='lilly')
```

### Return Values

#### Function Returning Single Values

```python
def get_name(first_name, last_name):
    full_name = f"{first_name} {last_name}"
    return full_name.title()
```

```python
x = get_name('mahmoud', 'hassan')
print(x)
```

```python
print(get_name('mahmoud', 'hassan'))
```

#### Function Returning Multiple Values

```python
def divide(a, b):
    quotient = a // b
    remainder = a % b
    return quotient,remainder
```

```python
q,r = divide(10, 3)
print(q)
print(r)
```

### Making an Argument Optional

```python
def get_name(first_name, last_name, middle_name=''):
    if middle_name != '':
        full_name = f"{first_name} {last_name} {middle_name}"
    else:
        full_name = f"{first_name} {last_name}"
    return full_name.title()
```

```python
name = get_name('mahmoud', 'hassan')
print(name)
```

```python
name= get_name('mahmoud', 'hassan', 'mohamed')
print(name)
```

## Exceptions and Error Handling

- Using `try` and `except` blocks for error handling  
- try block is used to wrap code that might raise an exception  
- except block handles the error gracefully

### Syntax

```python
print(5 / 0)
```

```python
try:
    print(5 / 0)  # This raises ZeroDivisionError
except ZeroDivisionError:
    print("You can't divide by zero!")
```

### ELSE Block

```python
try:
    answer = 10 / 2  # No error
except:
    print("You can't divide by zero!")
else:
    print(answer)
```

### PASS keyword allows the program to fail silently.

```python
try:
    print(10/0)
except ZeroDivisionError:
    pass  # The program will continue without errors
```

### Multiple Exceptions

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print(result)
except ValueError:
    print("Invalid input! Please enter a number.")
except ZeroDivisionError:
    print("You can't divide by zero!")
```

### Raising Custom Exceptions

```python
def withdraw(amount):
    if amount > 1000:
        raise ValueError("Withdrawal limit exceeded!")  # Custom error message
    print(f"Withdrawing {amount} dollars")

try:
    withdraw(500)
except ValueError as e:
    print(f"Transaction failed: {e}")
```

```python
withdraw(1500)
```

## Practical Examples

### Calculating Average of Numbers Using Functions

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

numbers = [10, 20, 30, 40, 50]
print("Average:", calculate_average(numbers))
```

### Finding the Longest Word in a List

```python
def find_longest_word(words):
    longest = ""
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest

words = ["apple", "banana", "cherry", "date"]
print("Longest word:", find_longest_word(words))
```

### Generating a List of Squares Using List Comprehension and Functions

```python
def square_list(numbers):
    return [x ** 2 for x in numbers]

numbers = [1, 2, 3, 4, 5]
print("List of squares:", square_list(numbers))
```

### Converting a List of Strings to Uppercase

```python
def convert_to_uppercase(words):
    list = []
    for word in words:
        list.append(word.upper())
    return list


words = ["hello", "world", "python", "functions"]
print("Uppercase words:", convert_to_uppercase(words))
```

### Finding the Minimum and Maximum Values in a List

```python
def find_min_max(numbers):
    return min(numbers), max(numbers)

numbers = [3, 7, 2, 9, 4]
min_value, max_value = find_min_max(numbers)
print("Minimum value:", min_value)
print("Maximum value:", max_value)
```
