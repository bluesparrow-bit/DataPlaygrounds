### Q_4: Add a new column FamilySize to the DataFrame, calculated as SibSp + Parch + 1 (including the passenger themselves)

```python
df['FamilySize'] = df.SibSp + df.Parch + 1
df.head()
```

#### Q_5: Fill missing values in the Age column with the mean age then Group the data by Pclass and calculate the average age for each class

```python
df.fillna({'Age':df.Age.mean()},inplace=True)
df.groupby('Pclass')['Age'].mean()
```

#### Q_Bonus: Calculate the survival rate (mean of Survived) for each gender (Sex)

```python
df.groupby(['Sex','Pclass'])['Survived'].mean().unstack() * 100
```

### Q_6: Split the Titanic DataFrame df into two parts: df1 (first 500 rows) and df2 (remaining rows). Concatenate them vertically to recreate the original DataFrame.

```python
df1 = df.loc[:499,:]
df2 = df.loc[499:,:]
df3 = pd.concat([df1,df2])
df3.info()
```

### Q_7: Create a new DataFrame passenger_details with columns PassengerId and Nationality. Perform a left join to merge it with df, keeping all original passengers.

```python
passenger_details = pd.DataFrame(columns=['Nationality'])
passenger_details['Passengerid'] = df.PassengerId 
left_join = df.join(passenger_details,how='left').drop(columns='Passengerid')
left_join.head()
```

### Q_8 :  Group the data by Pclass and Sex, then calculate the average Age and total Fare for each group.

```python
df.groupby(['Pclass','Sex']).aggregate({'Age':'mean','Fare':'sum'})
```

```python
s_age = df.groupby(['Pclass','Sex'])['Age'].mean()
s_fare = df.groupby(['Pclass','Sex'])['Fare'].sum()
s_fare.rename('Total_Fare',inplace=True)
pd.concat([s_age,s_fare],axis=1)
```
