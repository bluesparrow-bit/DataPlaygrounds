## Compare List, Tuples, Dictionary

| Feature                      | **List**                           | **Tuple**                               | **Dictionary**                           |
| ---------------------------- | ---------------------------------- | --------------------------------------- | ---------------------------------------- |
| **Syntax**                   | `[]`                               | `()`                                    | `{}`                                     |
| **Data Storage**             | Stores values                      | Stores values                           | Stores key-value pairs                   |
| **Ordered**                  | ✅ Yes                              | ✅ Yes                                   | ✅ Yes (Python 3.7+)                      |
| **Mutable (Can be changed)** | ✅ Yes                              | ❌ No                                    | ✅ Yes                                    |
| **Indexed**                  | ✅ Yes                              | ✅ Yes                                   | ❌ No (access by key)                     |
| **Access Method**            | By index                           | By index                                | By key                                   |
| **Duplicate Values**         | ✅ Allowed                          | ✅ Allowed                               | Values: ✅ Allowed<br>Keys: ❌ Not allowed |
| **Can Add Items**            | ✅ Yes                              | ❌ No                                    | ✅ Yes                                    |
| **Can Remove Items**         | ✅ Yes                              | ❌ No                                    | ✅ Yes                                    |
| **Can Update Items**         | ✅ Yes                              | ❌ No                                    | ✅ Yes                                    |
| **Search Speed**             | Moderate                           | Moderate                                | Fast (by key)                            |
| **Memory Usage**             | Higher                             | Lower than lists                        | Higher (stores keys and values)          |
| **Best Used For**            | Collections that change frequently | Fixed collections that shouldn't change | Data with named attributes or mappings   |
| **Example Use Case**         | Shopping list, student names       | Coordinates, RGB color, days of week    | Student record, employee information     |

## while Loops

### Syntax

```python
number = 1
while number <= 5:
    print(number)
    number += 1
```

```python
prompt = "Tell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program."
message = ""
while message != 'quit':
    message = input(prompt)
    print(message)
```

### Using a Flag

```python
prompt = "Tell me something, and I will repeat it back to you:"
prompt += "\nEnter quit to end the program. "
active = True
while active:
    message = input(prompt)
    if message == 'quit':
        active = False
    else:
        print(message)
```

### Using break to Exit a Loop

```python
prompt = "Please enter the name of a city you have visited:"
prompt += "\n(Enter quit when you are finished.)"
while True:
    city = input(prompt)
    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
```

You can use the break statement in any of Python’sloops. 
- For example, you could use break to quit a for loop that’s working through a list or a dictionary

### Using continue in a Loop

```python
current_number = 0
while current_number < 10:
    current_number +=1
    if current_number %2==0:
        continue
    print(current_number)
```

### Calculating Factorial of a Number

```python
num = 5
factorial = 1

for i in range(1, num + 1):
    factorial *= i

print("Factorial of", num, "is", factorial)
```

### Validating User Input with While Loop

```python
while True:
    user_input = input("Enter a number greater than 10: ")
    if user_input.isdigit() and int(user_input) > 10:
        print("Thank you!")
        break
    else:
        print("Invalid input. Please try again.")
```

### Drawing a Simple Pattern Using Loops

```python
rows = 5

for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print("")
```

### Creating a Dictionary from Two Lists

```python
keys = ["name", "age", "location"]
values = ["Alice", 30, "New York"]
dictionary = {}

for i in range(len(keys)):
    dictionary[keys[i]] = values[i]

print(dictionary)
```

### Finding the Maximum Value in a List

```python
numbers = [3, 5, 2, 8, 1, 9, 4]
max_value = numbers[0]

for number in numbers:
    if number > max_value:
        max_value = number

print("The maximum value is:", max_value)
```
