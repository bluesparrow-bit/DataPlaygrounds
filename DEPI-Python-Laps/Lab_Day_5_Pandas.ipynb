## Series

- A **`Series`** is a one-dimensional sequence of data values.
- If a **`DataFrame`** is like a table, a **`Series`** is like a list or a single column.
- A `Series` can be created directly from a Python list.
- Each value in a `Series` has an associated **index**.

```python
pd.Series([1, 2, 3, 4, 5])
```

- A **`Series`** can be thought of as a **single column of a DataFrame**.
- Like a DataFrame, a Series can have custom **row labels** using the `index` parameter.
- A Series does **not** have a column name.
- Instead, a Series has one overall **`name`**.
- The `name` identifies the Series itself.

```python
pd.Series([30, 35, 40], index=['2015 Sales', '2016 Sales', '2017 Sales'], name='Product A')
```

- A **`Series`** and a **`DataFrame`** are closely related in Pandas.
- A **DataFrame can be viewed as a collection of Series**.
- Each **column in a DataFrame is a Series**.
- Multiple Series are combined together to form a DataFrame.

## Indexing in pandas

`loc` and `iloc`

- Both **`loc`** and **`iloc`** use the same order:
  - **Row first**
  - **Column second**
- This is different from standard Python indexing, which is generally **column first, row second** when working with nested structures.
- With `loc` and `iloc`, the basic structure is:

### iloc

#### Using `:` with `iloc`

- The **`:`** operator means **"select everything"**.
- When used with a range, it means **select values within that range**.
- With `iloc`, the syntax is:

```python
reviews.iloc[:, 0]
```

```python
reviews.iloc[:3, 0]
```

```python
reviews.iloc[1:3, 0]
```

```python
reviews.iloc[-5:]
```

#### It's also possible to pass a list

```python
reviews.iloc[[0, 1, 2], 0]
```

### loc

#### Using `:` with `loc`

```python
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```

#### Label-based selection

```python
reviews.loc[0, 'country']
```

### Comparison Between `loc` and `iloc`

| Feature            | `loc`                  | `iloc`                    |
| ------------------ | ---------------------- | ------------------------- |
| Uses               | Labels                 | Integer positions         |
| Rows               | Labels                 | Positions                 |
| Columns            | Labels                 | Positions                 |
| String labels      | ✅ Yes                  | ❌ No                      |
| Integer positions  | ❌ Not directly         | ✅ Yes                     |
| Slice end          | **Included**           | **Excluded**              |
| Boolean conditions | ✅ Yes                  | Limited                   |
| Negative positions | ❌ No                   | ✅ Yes                     |
| Custom indexes     | ✅ Very useful          | ✅ Still works by position |
| Select last row    | `df.loc[df.index[-1]]` | `df.iloc[-1]`             |


## Missing data

###
- Missing values in Pandas are commonly represented by **`NaN`**.
- **`NaN`** stands for **"Not a Number"**.
- `NaN` is commonly used to represent **missing or unavailable data**.
- For technical reasons, traditional `NaN` values are stored as **`float64`**.

```python
reviews[pd.isnull(reviews.country)].head()
```

```python
reviews[reviews.country.isnull()].head()
```

###
- Replacing missing values is a common task when working with data.
- Pandas provides the **`fillna()`** method to replace missing values.
- `fillna()` allows us to use different strategies for handling `NaN` values.
- One simple approach is to replace every `NaN` with a specific value such as **`"Unknown"`**.

```python
reviews.region_2.fillna("Unknown")
```

###
- Sometimes we need to replace a value that is **not missing (`NaN`)**.
- Pandas provides the **`replace()`** method for this purpose.
- `replace()` can be used to change an existing value into a new value.
- It is useful when correcting or updating data.

```python
reviews.taster_twitter_handle.replace("@kerinokeefe", "@kerino")
```
