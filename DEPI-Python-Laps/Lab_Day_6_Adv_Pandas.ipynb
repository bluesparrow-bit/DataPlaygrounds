## Concate And Grouping

### Creating a DataFrame for Examples

- A custom function can be created to **generate a DataFrame** in a specific format.
- This makes it easier to create **consistent sample data** for examples.
- The function can be reused multiple times throughout the examples.
- It helps simplify demonstrations of Pandas operations.

````python
def make_df(cols, ind): 
    """Quickly create a DataFrame.""" 
    data = {}
    for col in cols: 
        values = [] 
        for i in ind: 
            values.append(str(col) + str(i)) 
        data[col] = values 
    return pd.DataFrame(data, index=ind)

make_df('ABC', range(3))
````

````python
class display(object):
    """Display HTML representation of multiple objects"""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    def __init__(self, *args):
        self.args = args
        
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                        for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                        for a in self.args)
````

### PD Concate: 

- `pd.concat()` is used to **concatenate (combine) multiple `Series` or `DataFrame` objects**.
- It works similarly to NumPy's `np.concatenate()`.
- `pd.concat()` provides additional options that give more control over how data is combined.

#### Basic Syntax

```python
pd.concat(
    [obj1,obj2,..],
    axis=0,
    join='outer',
    ignore_index=False,
    keys=None,
    levels=None,
    names=None,
    verify_integrity=False,
    sort=False,
    copy=True
)

#### Important Parameters

- **`objs`** → The `Series` or `DataFrame` objects to combine.
- **`axis=0`** → Combine objects **row-wise (vertically)** | **`axis=1`** → Combine objects **column-wise (horizontally)**.
- **`join='outer'`** → Include all indexes or columns from the objects | **`join='outer'`** → Include common columns only.
- **`ignore_index=False`** → Keep the original indexes | **`ignore_index=True`** → Create a new sequential index.
- **`keys`** → Add labels to identify the original objects.
- **`sort=False`** → Keep the original column order.
- **`verify_integrity=False`** → Do not check for duplicate indexes | Set to **`True`** to check for duplicate indexes.

## Merge and Join

Relational Algebra

- **Relational Algebra** is a formal set of rules used to manipulate **relational data**.
- It is the conceptual foundation of many operations used in **databases**.
- It provides fundamental operations that can be combined to perform more complex data operations.
- Pandas implements several relational algebra operations through:
- **`pd.merge()`**
- **`join()`** method of `Series` and `DataFrame`
- These operations allow you to **combine and link data from different sources efficiently**.

### Merge

Categories of Merge

- `pd.merge()` supports **three main types of joins**:
  - **One-to-One**
  - **Many-to-One**
  - **Many-to-Many**
- All three types use the same `pd.merge()` function.
- The type of join depends on the **relationship between the data in the input DataFrames**.
- The basic syntax is: pd.merge(df1, df2)

#### One-to-One Merge

How `pd.merge()` Works

- `pd.merge()` automatically identifies the **common column** between the two DataFrames.
- In this example, both DataFrames contain an **`employee`** column.
- The `employee` column is used as the **key** to match the rows.
- The result is a **new DataFrame** containing information from both DataFrames.

Important Points

- The order of rows does **not need to be the same** in both DataFrames.
- `pd.merge()` matches rows based on the **key values**, not their positions.
- By default, the original **index is discarded** after merging.
- An exception occurs when merging based on the index using:
  - **`left_index=True`**
  - **`right_index=True`**

Main Idea

> `pd.merge()` combines DataFrames by matching values in a common key column, regardless of the original row order.

````python
df1 = pd.DataFrame({'employee': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'group': ['Accounting', 'Engineering',
                              'Engineering', 'HR']})
df2 = pd.DataFrame({'employee': ['Lisa', 'Bob', 'Jake', 'Sue'],
                    'hire_date': [2004, 2008, 2012, 2014]})
display('df1', 'df2')
````

````python
df3 = pd.merge(df1, df2)
df3
````

#### Many-to-One Merge

- A **many-to-one join** occurs when one of the two DataFrames contains **duplicate values** in the key column.
- Multiple rows from one DataFrame can match a **single row** in the other DataFrame.
- `pd.merge()` preserves the duplicate entries in the resulting DataFrame.
- The matching is performed using the **common key column**.

Main Idea

- **Many rows** → match **one row**.
- Duplicate key values are **preserved** in the result.
- `pd.merge()` automatically combines the matching rows.

````python
df4 = pd.DataFrame({'group': ['Accounting', 'Engineering', 'HR'],
                    'supervisor': ['Carly', 'Guido', 'Steve']})
display('df3', 'df4', 'pd.merge(df3, df4)')
````

#### Many-to-Many Merge

- A **many-to-many join** occurs when the key column contains **duplicate values in both DataFrames**.
- Each matching key in the left DataFrame can be combined with **all matching keys** in the right DataFrame.
- This can produce **multiple rows for each matching key**.
- The resulting number of rows can increase significantly because of the multiple combinations.
- Many-to-many joins are useful when multiple records are associated with the same key.

Example Concept

- A DataFrame contains **people and their groups**.
- Another DataFrame contains **groups and their associated skills**.
- A many-to-many merge can be used to find the **skills associated with each person**.

Main Idea

> **Many-to-many:** Multiple rows in the left DataFrame match multiple rows in the right DataFrame.

````python
df5 = pd.DataFrame({'group': ['Accounting', 'Accounting',
                              'Engineering', 'Engineering', 'HR', 'HR'],
                    'skills': ['math', 'spreadsheets', 'software', 'math',
                               'spreadsheets', 'organization']})
display('df1', 'df5', "pd.merge(df1, df5)")
````

### Specification of the Merge Key

#### Merge Keys

- By default, `pd.merge()` looks for **common column names** in both DataFrames.
- The matching column is automatically used as the **merge key**.
- Sometimes, the column names in the two DataFrames are **different**.
- In this case, `pd.merge()` provides options to specify the columns that should be used as the merge keys.

Main Idea

- **Same column names** → `pd.merge()` can automatically identify the key.
- **Different column names** → Specify the merge keys explicitly.
- The main options for handling different key names are:
  - **`on`**
  - **`left_on`**
  - **`right_on`**
  - **`left_index`**
  - **`right_index`**

````python
display('df1', 'df2', "pd.merge(df1, df2, on='employee')")
````

````python
df3 = pd.DataFrame({'name': ['Jake', 'Bob', 'Sue', 'Sue'],
                    'salary': [70000, 80000, 120000, 90000]})
display('df1', 'df3', 'pd.merge(df1, df3, left_on="employee", right_on="name").drop("name",axis=1)')
````

### Specifying Set Arithmetic for Joins

#### Set Arithmetic in Joins

- When performing a join, an important consideration is **which key values should be included** in the result.
- A key value may exist in one DataFrame but **not in the other DataFrame**.
- Pandas provides different types of joins to control how these unmatched values are handled.

#### Main Join Types

- **Inner Join** → Keep only matching values from both DataFrames.
- **Outer Join** → Keep all values from both DataFrames.
- **Left Join** → Keep all values from the left DataFrame.
- **Right Join** → Keep all values from the right DataFrame.

#### Main Idea

> The type of join determines **which matching and non-matching records are kept** in the final DataFrame.

### Overlapping Column Names: The suffixes Keyword

#### Handling Conflicting Column Names

- Sometimes, the two DataFrames being merged contain **columns with the same name**.
- This can cause a conflict because Pandas needs to distinguish between the two columns.
- `pd.merge()` automatically handles these conflicts by adding **suffixes** to the column names.
- By default, Pandas uses:
  - **`_x`** → Column from the left DataFrame.
  - **`_y`** → Column from the right DataFrame.

````python
df8 = pd.DataFrame({'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'rank': [1, 2, 3, 4]})
df9 = pd.DataFrame({'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'rank': [3, 1, 4, 2]})
display('df8', 'df9', 'pd.merge(df8, df9, on="name")')
````

````python
pd.merge(df8, df9, on="name", suffixes=["_L", "_R"])
````

### groupby: Split, Apply, Combine

Simple aggregations can give you a flavor of your dataset, but often we would prefer to aggregate conditionally on some label or index: this is implemented in the so-called `groupby` operation.
The name "group by" comes from a command in the SQL database language, but it is perhaps more illuminative to think of it in the terms first coined by Hadley Wickham of Rstats fame: *split, apply, combine*.

A canonical example of this split-apply-combine operation, where the "apply" is a summation aggregation, is illustrated in this figure:

![](split-apply-combine.png)

([figure source in Appendix](https://github.com/jakevdp/PythonDataScienceHandbook/blob/master/notebooks/06.00-Figure-Code.ipynb#Split-Apply-Combine))

This illustrates what the `groupby` operation accomplishes:

- The *split* step involves breaking up and grouping a `DataFrame` depending on the value of the specified key.
- The *apply* step involves computing some function, usually an aggregate, transformation, or filtering, within the individual groups.
- The *combine* step merges the results of these operations into an output array.

While this could certainly be done manually using some combination of the masking, aggregation, and merging commands covered earlier, an important realization is that *the intermediate splits do not need to be explicitly instantiated*. Rather, the `groupby` can (often) do this in a single pass over the data, updating the sum, mean, count, min, or other aggregate for each group along the way.
The power of the `groupby` is that it abstracts away these steps: the user need not think about *how* the computation is done under the hood, but rather can think about the *operation as a whole*.

As a concrete example, let's take a look at using Pandas for the computation shown in the following figure.
We'll start by creating the input `DataFrame`:

### Aggregate, Filter, Transform, Apply

GroupBy Operations

- `groupby()` provides several operations for working with grouped data.
- In addition to **aggregation**, `GroupBy` objects support:
  - **`aggregate()`** → Apply one or more aggregation functions to each group.
  - **`filter()`** → Keep or remove groups based on a condition.
  - **`transform()`** → Perform calculations on each group and return results with the original shape.
  - **`apply()`** → Apply a custom function to each group.
- These operations allow us to perform different calculations **before combining the grouped data**.

Main Idea

> `GroupBy` provides several methods to efficiently process and analyze data **group by group**.

Common `GroupBy` Methods

| Method | Purpose |
|---|---|
| `aggregate()` | Calculate summary statistics for each group |
| `filter()` | Filter groups based on a condition |
| `transform()` | Transform values while keeping the original shape |
| `apply()` | Apply a custom function to each group |

- The following examples will use a specific `DataFrame` to demonstrate these operations.

````python
rng = np.random.RandomState(0)
df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'data1': range(6),
                   'data2': rng.randint(0, 10, 6)},
                   columns = ['key', 'data1', 'data2'])
df
````

#### aggregate

````python
df.groupby('key').aggregate(['min', np.median, max])
````

````python
df.groupby('key').aggregate({'data1': 'min',
                             'data2': 'max'})
````

#### Filtering

````python
def filter_func(x):
    return x['data2'].std() > 4

display('df', "df.groupby('key').std()",
        "df.groupby('key').filter(filter_func)")
````

#### Transformation

````python
df
````

````python
def center(x):
    return x - x.mean()
df.groupby('key').transform(center)
````

#### apply

````python
df
````

````python
def norm_by_data2(x):
    # x is a DataFrame of group values
    x['data1'] /= x['data2'].sum()
    return x

df.groupby('key').apply(norm_by_data2)
````

### Grouping Example

As an example of this, in a few lines of Python code we can put all these together and count discovered planets by method and by decade:

````python
decade = 10 * (planets['year'] // 10)
decade = decade.astype(str) + 's'
decade.name = 'decade'
planets.groupby(['method', decade])['number'].sum().unstack().fillna(0)
````

This shows the power of combining many of the operations we've discussed up to this point when looking at realistic datasets: we quickly gain a coarse understanding of when and how extrasolar planets were detected in the years after the first discovery.
