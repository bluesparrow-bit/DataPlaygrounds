### Priority Queue

```python
from queue import PriorityQueue

pq = PriorityQueue()

pq.put((2, "Bob (Broken Arm)"))      # Priority 2
pq.put((1, "Alice (Heart Attack)"))  # Priority 1 (Higher priority)
pq.put((3, "Charlie (Mild Fever)"))  # Priority 3 (Lowest priority)

pq
```

```python
while not pq.empty():
    priority, patient = pq.get()
    print(f"Treating: {patient}")
```

### Linked Lists

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def __repr__(self):
        return self.data

class LinkedList:
    def __init__(self):
        self.head = None

    def __repr__(self):
        node = self.head
        nodes = []
        while node is not None:
            nodes.append(node.data)
            node = node.next
        nodes.append("None")
        return " -> ".join(nodes)
```

```python
node2 = Node('b')
node3 = Node('c')
node1.next = node2
node2.next = node3
llist
```

Create linked list with predefined data, Traverse and iterate a linked list.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def __repr__(self):
        return self.data

class LinkedList:
    def __init__(self, nodes = None):
        self.head = None
        if nodes is not None:
            node = Node(data=nodes.pop(0))
            self.head = node
            for elem in nodes:
                node.next = Node(data=elem)
                node = node.next

    def __repr__(self):
        node = self.head
        nodes = []
        while node is not None:
            nodes.append(node.data)
            node = node.next
        nodes.append("None")
        return " -> ".join(nodes)
    def __iter__(self):
        node = self.head
        while node is not None:
            yield node
            node = node.next
```

```python
def add_after(self, target_node_data, new_node):
    if self.head is None:
        raise Exception("List is empty")

    for node in self:
        if node.data == target_node_data:
            new_node.next = node.next
            node.next = new_node
            return

    raise Exception("Node with data '%s' not found" % target_node_data)
LinkedList.add_after = add_after
```

##### Implement the `add_before` method
*Hint:* Use the function heading: `def add_before(self, target_node_data, new_node)`

```python
# solution 
def add_before(self, target_node_data, new_node):
    if self.head is None:
        raise Exception("List is empty")

    if self.head.data == target_node_data:
        return self.add_first(new_node)

    prev_node = self.head
    for node in self:
        if node.data == target_node_data:
            prev_node.next = new_node
            new_node.next = node
            return
        prev_node = node

    raise Exception("Node with data '%s' not found" % target_node_data)
```

#### Remove node

```python
def remove_node(self, target_node_data):
    if self.head is None:
        raise Exception("List is empty")

    if self.head.data == target_node_data:
        self.head = self.head.next
        return

    previous_node = self.head
    for node in self:
        if node.data == target_node_data:
            previous_node.next = node.next
            return
        previous_node = node

    raise Exception("Node with data '%s' not found" % target_node_data)
LinkedList.remove_node = remove_node
```

##### Using The Hash Tables Practical Applications

```python
gradebook = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78
}

print(f"Alice's grade: {gradebook['Alice']}")

gradebook["David"] = 88  

gradebook["Bob"] = 95  
```

```python
for student in gradebook:
    print(student)
```

```python
for grade in gradebook.values():
    print(grade)
```

```python
for student, grade in gradebook.items():
    print(f"{student}: {grade}")
```
