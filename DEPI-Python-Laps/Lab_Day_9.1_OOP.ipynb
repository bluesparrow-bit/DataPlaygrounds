### Polymorphism

Changing the Function Behavior Based on Types

```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old"

    def activity(self):
        return f"{self.name} enjoys playing and being a good companion."
```

```python
class Labrador(Dog):

    def activity(self):
        return f"{self.name} loves swimming and retrieving objects!"


class Beagle(Dog):
    
    def activity(self):
        return f"{self.name} enjoys tracking scents and following trails!"
```

```python
Buddy = Labrador("Buddy", 3)
print(Buddy.activity())
```

```python
Charlie = Beagle("Charlie", 2)
print(Charlie.activity())
```

### Abstraction

The OOP paradigm allows creating a base class without much functionality but a mutual behavior for children classes.
- You can list the animal attributes in the class `Animal` then introduce values based on the animal type for the child class.
- You need to import `abc` and `abstractmethod`.
- You can use the interfaces to inherit from multiple classes.

```python
from abc import ABC, abstractmethod # Abstract Base Class

# Abstract Class: Animal
class Animal(ABC):
    def __init__(self, name):
        self.name = name

    # Abstract method to be implemented by subclasses
    @abstractmethod
    def make_sound(self):
        pass
```

```python
class Swimmer(ABC):
    # Abstract method to be implemented by classes
    @abstractmethod
    def swim(self):
        pass
```

```python
# Concrete Class: Dog (inherits from Animal and implements Swimmer)
class Dog(Animal, Swimmer):

    def make_sound(self):
        return "Woof!"

    def swim(self):
        return f"{self.name} is swimming."

dog = Dog(name="Buddy")

print(f"{dog.name} says: {dog.make_sound()}")
print(f"{dog.name} can swim: {dog.swim()}")
```

```python
# Concrete Class: Cat (inherits from Animal)
class Cat(Animal):
    
    def make_sound(self):
        return "Meow!"

cat = Cat(name="Whiskers")
print(f"{cat.name} says: {cat.make_sound()}")
```

Inheritance vs Abstraction

| Inheritance | Abstraction |
|---|---|
| Focuses on **reusing code** | Focuses on **hiding complexity** |
| Parent → Child relationship | Defines a common interface/rule |
| Child gets parent's functionality | Child must implement required functionality |
| Usually uses `class Child(Parent)` | Often uses `ABC` + `@abstractmethod` |

### Encapsulation

Hiding Properties

**Encapsulation** is an OOP concept that restricts direct access to an object's data and methods.

It helps us:

- Protect internal data
- Control how data is modified
- Prevent unwanted changes
- Ensure data is used correctly

Encapsulation is commonly implemented using **private attributes**, **getters**, and **setters**.

```python
class Dog:
    def __init__(self, name, age, breed):
        self.name = name    # Public Attribute
        self.age = age      # Public Attribute
        self.__breed = breed  # Private Attribute (Encapsulated)

    def get_breed(self):
        """Getter method to access the private breed attribute"""
        return self.__breed
```

```python
dog = Dog("Buddy", 3, "Golden Retriever")
print(dog.name)
print(dog.age)

#print(dog.__breed)
```

```python
print(dog.get_breed())
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.__age = age  # Private Attribute

    def get_age(self):
        """Getter method to retrieve the dog's age"""
        return self.__age

    def set_age(self, new_age):
        """Setter method to ensure age is valid before updating"""
        if new_age > 0:
            self.__age = new_age
        else:
            print("Age must be positive!")
```

```python
dog = Dog("Charlie", 5)
print(dog.get_age())
```

```python
dog.__age = 7
```

```python
print(dog.__age)
```

```python
print(dog.get_age())
```

```python
dog.set_age(10)
```

```python
print(dog.__age)
```

```python
print(dog.get_age())
```
